# HW Redemption Grades

```python
from redemption_grades import process_redemption

# Change HW for later assignments, e.g. "hw02"
HW = "hw03"

# Update the paths to match (most likely will need to change just CANVAS_CSV)
CANVAS_CSV = "data/2026-11-01T1059_Grades-DSC020_FA26_001.csv"
ORIGINAL_CSV = f"data/{HW}_scores.csv"
CHECKPOINT_CSV = f"data/{HW}_checkpoint_scores.csv" # Set to None if no checkpoint (see next line)
# CHECKPOINT_CSV = None
REDEMPTION_CSV = f"data/{HW}_redemption_scores.csv"
OUTPUT_CSV = f"data/canvas_{HW}_updated.csv"

result = process_redemption(
    hw=HW,
    canvas_csv=CANVAS_CSV,
    original_csv=ORIGINAL_CSV,
    redemption_csv=REDEMPTION_CSV,
    checkpoint_csv=CHECKPOINT_CSV, 
    output_csv=OUTPUT_CSV,
)
```

## Summary

```python
print("Assignment column:", result["hw_col"])
print("Max points:", result["max_points"])
print(f"Checkpoint max points: {result['checkpoint_max_points']}")
print(f"Checkpoint submissions: {len(result['submitted_checkpoint'])}")
print("Redemption increased score:", result["increased_count"])
print("Redemption did not increase score:", result["not_increased_count"])
print("Canvas students matched:", result["matched_count"])
print("Gradescope accounts not found in Canvas:", len(result["unmatched"]))
print("Saved:", result["output_csv"].resolve())
```
